# SheAlert: Passive Women's Safety App

## Problem Statement
Women in unsafe situations often **cannot manually trigger** an alert. This project builds a system that detects danger **passively** using AI.

## Core Objectives
- Detect distress sounds using **audio classification**
- Detect abnormal movement using **motion sensors**
- Verify threats using the **camera**

### Environment Requirements
- Python 3.10+
- Libraries: numpy, opencv-python, matplotlib, ultralytics, torch
